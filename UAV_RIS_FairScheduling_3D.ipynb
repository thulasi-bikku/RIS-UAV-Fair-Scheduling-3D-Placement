```python
# FAIR-RIS-UAV reproduction -- paste each STEP block into its own Colab cell (or the whole file into one cell)



# ============================================================================
# # FAIR-RIS-UAV — full reproduction
# **Fairness-Aware Scheduling and Joint 3D Placement for RIS-Assisted Dual-Hop UAV Relaying over Mixed α–κ–μ and Rayleigh Fading Channels**
#
# Run the cells top to bottom (`Runtime → Run all`). Each step prints the numbers quoted in the paper next to the values computed here, and every figure is saved as a 300-dpi PNG in `figures/`.
#
# | Step | What it produces |
# |---|---|
# | 1–2 | Imports, Table 1 parameters, geometry |
# | 3–5 | Channel model, Eqs. (1)–(7) |
# | 6 | Closed-form outage (9)–(10) and ASER (13) |
# | 7 | Monte Carlo simulator (element-by-element RIS) |
# | 8 | Algorithm 2 – multi-start 3D placement |
# | 9–16 | Fig. 2, Fig. 5, Fig. 3, Table 2, Fig. 4, Fig. 6, Table 3, Fig. 7, Fig. 8 |
#
# Runtime on a standard Colab CPU: roughly 6–10 minutes in total (Fig. 5 and Fig. 8 are the slowest).
# ============================================================================


# ============================================================================
# ## Step 1 — Imports and plot style
# ============================================================================

import numpy as np, os, time
import matplotlib.pyplot as plt
from scipy import special as sp, stats
from scipy.optimize import minimize
from scipy.integrate import quad

os.makedirs('figures', exist_ok=True)
plt.rcParams.update({
    'font.family': 'serif', 'font.serif': ['DejaVu Serif'], 'mathtext.fontset': 'dejavuserif',
    'font.size': 10, 'axes.grid': True, 'grid.alpha': 0.35, 'legend.fontsize': 8.5,
    'figure.dpi': 110, 'savefig.dpi': 300, 'savefig.bbox': 'tight'})
C_FAIR, C_MAX, C_RR = 'tab:red', 'tab:blue', 'tab:gray'
C_M = {0: 'tab:gray', 16: 'tab:blue', 32: 'tab:red', 64: 'tab:green'}


# ============================================================================
# ## Step 2 — Simulation parameters (Table 1) and geometry
# All distances are normalised to 100 m inside the path-loss terms. The BS-to-UAV power offset is $\rho_S/\rho = +10$ dB, target rate $R = 1$ bps/Hz, so $\gamma_{th} = 2^{2R}-1 = 3$.
# ============================================================================

U = 100.0                      # distance normalisation (m)
alpha, mu = 2.5, 1.5           # first-hop alpha-kappa-mu
ae, be = 4.88, 0.43            # suburban A2G constants [5]
bN, bL, bd = 3.5, 2.0, 3.0     # NLoS / LoS PLE (hop 1) and direct second-hop PLE
k0, k90 = 10**(5/10), 10**(15/10)   # Rician factor at 0 and pi/2 (5 dB, 15 dB)
xi = 0.002                     # RIS gain constant (-27 dB)
hb = 30.0                      # BS height (m)
off_S = 10.0                   # rho_S / rho in dB
R_t = 1.0
g_th = 2**(2*R_t) - 1          # = 3 (4.77 dB)

users = np.array([[540, 60], [650, -80], [700, 40], [590, -30]], float)   # D1..D4 (ground)
ris = np.array([600, 140, 20.0])
N = len(users)
P_OPT = (314, 53, 222)         # optimised UAV position (M=32, 20 dB) from Algorithm 2
LO, HI = np.array([-100, -200, 50.]), np.array([800, 200, 300.])
print(f'N = {N}, gamma_th = {g_th} ({10*np.log10(g_th):.2f} dB), alpha*mu/2 = {alpha*mu/2}')


# ============================================================================
# ## Step 3 — First hop: elevation-dependent α–κ–μ (Eqs. 1–3)
# $P_{LoS}$ uses θ in degrees; κ(θ) uses θ in radians. $X$ is a unit-mean α–κ–μ power gain; $Q_\mu$ is evaluated through the non-central χ² CDF: $1-Q_\mu(a,b)=F_{\chi'^2_{2\mu}(a^2)}(b^2)$.
# ============================================================================

def hop1(p):
    x, y, h = p
    r1 = np.hypot(x, y)
    th = np.arctan2(h - hb, r1)                         # elevation (rad)
    d_SR = np.sqrt(r1**2 + (h - hb)**2) / U
    PL = 1 / (1 + ae*np.exp(-be*(np.degrees(th) - ae)))
    beta = bN + (bL - bN)*PL
    kap = k0*(k90/k0)**(2*th/np.pi)
    return d_SR, beta, kap

def xhat(kap):
    num = sp.gamma(mu + 2/alpha)*np.exp(-kap*mu)*sp.hyp1f1(mu + 2/alpha, mu, kap*mu)
    den = sp.gamma(mu)*(mu*(1 + kap))**(2/alpha)
    return den/num

def F_X(x, kap):
    b = np.sqrt(2*mu*(1 + kap))*(x/xhat(kap))**(alpha/4)
    return stats.ncx2.cdf(b**2, 2*mu, 2*kap*mu)

# sanity check: E[X] = 1
kap = hop1(P_OPT)[2]; rng = np.random.default_rng(1)
V = rng.noncentral_chisquare(2*mu, 2*kap*mu, 10**6)
print('kappa at optimum = %.2f,  MC E[X] = %.4f' % (kap, (xhat(kap)*(V/(2*mu*(1+kap)))**(2/alpha)).mean()))


# ============================================================================
# ## Step 4 — Second hop: direct link + RIS path, Gamma approximation (Eqs. 4–7)
# $\ell_{d,n}=d_n^{-\beta_d}$, $\ell_{r,n}=\xi/(d_{UI}^2 d_{In}^2)$; the amplitude $Z_n$ is moment-matched to a Gamma law with $k_n=\mu_n^2/\sigma_n^2$, $\vartheta_n=\sigma_n^2/\mu_n$. For $M=0$ the exact exponential CDF is used.
# ============================================================================

def gains(p):
    up = np.asarray(p, float)
    d_n  = np.sqrt(((users - up[:2])**2).sum(1) + up[2]**2) / U
    d_UI = np.linalg.norm(up - ris) / U
    d_In = np.sqrt(((users - ris[:2])**2).sum(1) + ris[2]**2) / U
    return d_n**(-bd), xi/(d_UI**2 * d_In**2)

def F_n(g, rho, ld, lr, M):
    if M == 0:
        return 1 - np.exp(-g/(rho*ld))
    m = np.sqrt(np.pi)/2*(np.sqrt(ld) + M*np.sqrt(lr))
    v = (1 - np.pi/4)*(ld + M*lr)
    k, th = m**2/v, v/m
    return sp.gammainc(k, np.sqrt(g/rho)/th)        # regularised lower incomplete gamma

ld, lr = gains(P_OPT)
print('l_d,n =', np.round(ld, 4)); print('l_r,n =', np.array2string(lr, precision=3))


# ============================================================================
# ## Step 5 — Fair CDF-based scheduling (Eq. 8, Lemma 1)
# Each user reports $U_n = F_n(\gamma_n)$; the relay serves $\arg\max_n U_n$. This is implemented inside the Monte Carlo simulator (Step 7). Quick check of Lemma 1 with i.i.d. uniforms:
# ============================================================================

Uq = np.random.default_rng(0).random((10**6, N))
print('access shares :', np.bincount(Uq.argmax(1), minlength=N)/10**6)
print('Pr{U_max<=0.5} =', (Uq.max(1) <= 0.5).mean(), ' vs 0.5^N =', 0.5**N)


# ============================================================================
# ## Step 6 — Closed-form outage (Theorem 1, Eqs. 9–10) and ASER (Eq. 13)
# $\gamma_1 = aX$ with $a=\rho_S/d_{SR}^{\beta(\theta_1)}$. BPSK: $(A,B)=(1,0.5)$, $\bar P_s=\frac{A^B}{2\Gamma(B)}\int_0^\infty e^{-A\gamma}\gamma^{B-1}F_{\gamma_{SD}}(\gamma)d\gamma$.
# ============================================================================

def pout(p, rho_dB, M, sched='fair', g=g_th, per_user=False):
    rho, rS = 10**(rho_dB/10), 10**((rho_dB + off_S)/10)
    d, beta, kap = hop1(p)
    P1 = F_X(g/(rS/d**beta), kap)
    ld, lr = gains(p)
    F = F_n(g, rho, ld, lr, M)
    if sched == 'fair':
        pn = 1 - (1 - P1)*(1 - F**N)
    elif sched == 'rr':
        pn = 1 - (1 - P1)*(1 - F)
    elif sched == 'max':
        return 1 - (1 - P1)*(1 - np.prod(F))
    return pn if per_user else pn.mean()

def aser(p, rho_dB, M, sched='fair', A=1.0, B=0.5):
    f = lambda g: np.exp(-A*g)*g**(B-1)*pout(p, rho_dB, M, sched, g=g)
    return A**B/(2*sp.gamma(B))*quad(f, 0, np.inf, limit=200)[0]

def snr_at(target, M, sched='fair', p=P_OPT, grid=np.arange(5, 30.01, 2.5)):
    # SNR for a target outage, read off the 2.5-dB plotting grid (log-linear interpolation), as in the paper
    y = np.log10([pout(p, r, M, sched) for r in grid])
    return np.interp(np.log10(target), y[::-1], grid[::-1])

for M, ref in [(0, 0.471), (16, 0.0161), (32, 5.73e-6), (64, 4.22e-6)]:
    print(f'M={M:3d}: Pout(20 dB) = {pout(P_OPT, 20, M):.3e}   (paper {ref})')


# ============================================================================
# ## Step 7 — Monte Carlo simulator
# RIS amplitudes are generated **element by element**, so the Gamma approximation is tested rather than assumed. Per-user throughput is $\tau_n = R\cdot\Pr\{n^\star=n,\ \gamma_{SD}\ge\gamma_{th}\}$ (= $\frac{R}{N}(1-P_{out,n})$ under equal access), Jain's index from Eq. (12), ergodic throughput $\bar C=\tfrac12\mathbb E[\log_2(1+\gamma_{SD})]$.
# ============================================================================

def mc(p, rho_dB, M, n=200_000, seed=0, scheds=('fair', 'max', 'rr')):
    rng = np.random.default_rng(seed)
    rho, rS = 10**(rho_dB/10), 10**((rho_dB + off_S)/10)
    d, beta, kap = hop1(p)
    V = rng.noncentral_chisquare(2*mu, 2*kap*mu, n)
    g1 = rS/d**beta * xhat(kap)*(V/(2*mu*(1 + kap)))**(2/alpha)
    ld, lr = gains(p)
    s = 1/np.sqrt(2)                                   # unit-power Rayleigh
    Z = np.sqrt(ld)*rng.rayleigh(s, (n, N))
    if M > 0:
        G = np.zeros((n, N))
        for _ in range(M):
            G += rng.rayleigh(s, (n, N))
        Z += np.sqrt(lr)*G
    g2 = rho*Z**2
    idx_all = {'fair': F_n(g2, rho, ld, lr, M).argmax(1),     # Algorithm 1
               'max':  g2.argmax(1),
               'rr':   np.arange(n) % N}
    out = {}
    for sc in scheds:
        idx = idx_all[sc]
        gSD = np.minimum(g1, g2[np.arange(n), idx])
        ok = gSD >= g_th
        tau = np.array([R_t*np.mean((idx == k) & ok) for k in range(N)])
        out[sc] = dict(pout=1 - ok.mean(),
                       share=np.bincount(idx, minlength=N)/n,
                       tau=tau, J=(tau.sum()**2/(N*(tau**2).sum()) if tau.any() else 1/N),   # J -> 1/N when nothing gets through
                       C=0.5*np.mean(np.log2(1 + gSD)),
                       ser=np.mean(0.5*sp.erfc(np.sqrt(gSD))))        # BPSK
    return out

t = time.time(); r = mc(P_OPT, 15, 32)
print({k: round(v['pout'], 4) for k, v in r.items()}, f'({time.time()-t:.1f} s)')


# ============================================================================
# ## Step 8 — Algorithm 2: multi-start joint 3D placement (Eq. 14)
# Nelder–Mead on $J(\mathbf p)=\log_{10}\bar P^{fair}_{out}(\mathbf p)$ from 36 starting points, bounds enforced by a penalty.
# ============================================================================

def place(M, rho_dB, starts=None):
    def J(q):
        if np.any(q < LO) or np.any(q > HI):
            return 1e3 + np.sum(np.clip(LO - q, 0, None) + np.clip(q - HI, 0, None))
        return np.log10(max(pout(q, rho_dB, M), 1e-300))
    if starts is None:
        starts = [(x, y, h) for x in (0, 200, 400, 600) for y in (-100, 0, 100) for h in (80, 180, 280)]
    best = min((minimize(J, np.array(s, float), method='Nelder-Mead',
                         options=dict(xatol=1e-3, fatol=1e-6, maxiter=4000)) for s in starts),
               key=lambda r: r.fun)
    return best.x, 10**best.fun

placements = {}
for M, rho in [(0, 15), (0, 20), (32, 15), (32, 20)]:
    t = time.time(); x, po = place(M, rho); placements[(M, rho)] = (x, po)
    print(f'M={M:2d}, rho={rho} dB -> UAV* = ({x[0]:.0f}, {x[1]:.0f}, {x[2]:.0f}) m,  Pout = {po:.3e}  ({time.time()-t:.1f} s)')
print('Paper: (561,-2,161) 0.116 | (574,1,164) 3.25e-3 | (405,50,160) 2.88e-3 | (314,53,222) 5.73e-6')


# ============================================================================
# ## Step 9 — Fig. 2: outage vs SNR for M = 0, 16, 32, 64 (analysis + MC)
# ============================================================================

rho_line = np.linspace(5, 30, 101); rho_mc = np.arange(5, 30.01, 2.5)
fig, ax = plt.subplots(figsize=(5.6, 4.2))
for M in (0, 16, 32, 64):
    ax.semilogy(rho_line, [pout(P_OPT, r, M) for r in rho_line], color=C_M[M], label=f'M = {M} (analysis)')
    mcv = np.array([mc(P_OPT, r, M, n=200_000, scheds=('fair',))['fair']['pout'] for r in rho_mc])
    ax.semilogy(rho_mc[mcv > 0], mcv[mcv > 0], 'o', mfc='none', color=C_M[M])
ax.plot([], [], 'ko', mfc='none', label='Monte Carlo')
ax.set(ylim=(1e-6, 1.3), xlim=(4, 31), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='Average outage probability')
ax.legend(loc='lower left'); fig.savefig('figures/fig2_outage_vs_M.png'); plt.show()

ref = {0: 27.33, 16: 20.29, 32: 16.75, 64: 12.82}
for M in ref:
    s = snr_at(1e-2, M); print(f'M={M:2d}: SNR for 1e-2 = {s:.2f} dB (paper {ref[M]}), saving vs M=0 = {snr_at(1e-2,0)-s:.1f} dB')


# ============================================================================
# ## Step 10 — Fig. 5: (a) outage vs RIS size, (b) Jain's index vs RIS size at 15 dB
# ============================================================================

M_a = [0, 4, 8, 16, 24, 32, 48, 64, 96, 128]
M_b = [0, 8, 16, 32, 64, 128]
fig, (a, b) = plt.subplots(1, 2, figsize=(10, 3.9))
for rho, c in [(10, 'tab:blue'), (15, 'tab:green'), (20, 'tab:red')]:
    a.semilogy(M_a, [pout(P_OPT, rho, M) for M in M_a], 'o-', ms=4, color=c, label=fr'$\rho$ = {rho} dB')
a.set(xlabel='Number of RIS elements M', ylabel='Average outage probability', title='(a) Outage vs. RIS size')
a.legend()
Jb = {s: [] for s in ('fair', 'max', 'rr')}
for M in M_b:
    r = mc(P_OPT, 15, M, n=150_000)
    for s in Jb: Jb[s].append(r[s]['J'])
b.plot(M_b, Jb['fair'], 'o-', color=C_FAIR, ms=4, label='Proposed fair CDF-based')
b.plot(M_b, Jb['max'], 's-', color=C_MAX, ms=4, label='Max-SNR (conventional)')
b.plot(M_b, Jb['rr'], '^-', color=C_RR, ms=4, label='Round robin')
b.set(xlabel='Number of RIS elements M', ylabel=r"Jain's index at $\rho$ = 15 dB", title='(b) Fairness vs. RIS size')
b.legend(loc='center right'); fig.tight_layout(); fig.savefig('figures/fig5_ris_size.png'); plt.show()

print('Floor at 20 dB (M=128): %.2e (paper 4.22e-6); at 15 dB (M=64): %.2e (paper 1.54e-4)' % (pout(P_OPT,20,128), pout(P_OPT,15,64)))
print('Jain max-SNR M=0 -> 128: %.3f -> %.3f (paper 0.357 -> 0.250)' % (Jb['max'][0], Jb['max'][-1]))
print('Jain fair    M=0 -> 128: %.3f -> %.3f (paper 0.366 -> 1.000)' % (Jb['fair'][0], Jb['fair'][-1]))


# ============================================================================
# ## Step 11 — Fig. 3: (a) outage of the three schedulers, (b) Jain's index vs SNR (M = 32)
# ============================================================================

M = 32
fig, (a, b) = plt.subplots(1, 2, figsize=(10, 3.9))
sty = {'fair': (C_FAIR, 'o', 'Proposed fair CDF-based'), 'max': (C_MAX, 's', 'Max-SNR (conventional)'), 'rr': (C_RR, '^', 'Round robin')}
MC3 = {r: mc(P_OPT, r, M, n=200_000) for r in rho_mc}
for s, (c, mk, lab) in sty.items():
    a.semilogy(rho_line, [pout(P_OPT, r, M, s) for r in rho_line], color=c, label=lab)
    v = np.array([MC3[r][s]['pout'] for r in rho_mc]); a.semilogy(rho_mc[v > 0], v[v > 0], mk, mfc='none', color=c)
    b.plot(rho_mc, [MC3[r][s]['J'] for r in rho_mc], marker=mk, ms=4, color=c, label=lab)
a.set(ylim=(1e-6, 1.3), xlim=(4, 31), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='Average outage probability',
      title='(a) Outage (lines: analysis, markers: MC)'); a.legend(loc='lower left')
b.set(ylim=(0.2, 1.05), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel="Jain's fairness index (throughput)", title='(b) Throughput fairness')
fig.tight_layout(); fig.savefig('figures/fig3_schedulers.png'); plt.show()
for s, ref in [('fair', 16.75), ('max', 13.70), ('rr', 20.23)]:
    print(f'{s:4s}: SNR for 1e-2 = {snr_at(1e-2, 32, s):.2f} dB (paper {ref})')


# ============================================================================
# ## Step 12 — Table 2: scheduler comparison at ρ = 15 dB (M = 32, N = 4)
# ============================================================================

T2 = mc(P_OPT, 15, 32, n=300_000, seed=7)
rows = [('Access share D1/D2/D3/D4 (%)', lambda s: ' / '.join(f'{100*x:.1f}' for x in T2[s]['share'])),
        ('Average outage (MC)',         lambda s: f"{T2[s]['pout']:.3g}"),
        ('Weakest-user throughput',     lambda s: f"{T2[s]['tau'].min():.4g}"),
        ('Sum throughput (bps/Hz)',     lambda s: f"{T2[s]['tau'].sum():.3f}"),
        ("Jain's index J",              lambda s: f"{T2[s]['J']:.3f}"),
        ('SNR for Pout=1e-2 (dB)',      lambda s: f"{snr_at(1e-2, 32, s):.2f}")]
print(f"{'Metric':32s}{'Fair CDF-based':>26s}{'Max-SNR':>26s}{'Round robin':>26s}")
for name, f in rows:
    print(f"{name:32s}" + ''.join(f'{f(s):>26s}' for s in ('fair', 'max', 'rr')))
f, m, r = T2['fair'], T2['max'], T2['rr']
print(f"\nWeakest-user gain vs max-SNR: {f['tau'].min()/m['tau'].min():.0f}x (paper 37x); "
      f"sum-throughput loss: {100*(1-f['tau'].sum()/m['tau'].sum()):.1f}% (paper 12.6%)")
print(f"Analytical outage fair/RR: {pout(P_OPT,15,32):.3f} / {pout(P_OPT,15,32,'rr'):.3f} -> "
      f"reduction {100*(1-pout(P_OPT,15,32)/pout(P_OPT,15,32,'rr')):.0f}% (paper 67%, 0.125 vs 0.377)")


# ============================================================================
# ## Step 13 — Fig. 4: per-user throughput at ρ = 15 dB (M = 32)
# ============================================================================

fig, ax = plt.subplots(figsize=(5.6, 3.9)); w = 0.27; xk = np.arange(N)
for i, (s, (c, _, lab)) in enumerate(sty.items()):
    ax.bar(xk + (i-1)*w, T2[s]['tau'], w, color=c, label=lab)
ax.set_xticks(xk, [f'$D_{k+1}$' for k in range(N)]); ax.set_ylabel('Per-user throughput (bps/Hz)')
ax.grid(axis='x', visible=False); ax.legend(); fig.savefig('figures/fig4_per_user.png'); plt.show()


# ============================================================================
# ## Step 14 — Fig. 6: (a) outage map at h = 222 m, (b) outage vs altitude (20 dB)
# ============================================================================

xs, ys = np.linspace(-100, 800, 181), np.linspace(-200, 200, 81)
Zmap = np.array([[np.log10(pout((x, y, 222), 20, 32)) for x in xs] for y in ys])
hs = np.linspace(50, 300, 126); xo, yo = P_OPT[:2]; cl = (620, 0)

fig, (a, b) = plt.subplots(1, 2, figsize=(10.5, 4.2), gridspec_kw=dict(width_ratios=[1.15, 1]))
cf = a.contourf(xs, ys, np.clip(Zmap, -6.3, 0), levels=np.arange(-6.5, 0.01, 0.5), cmap='viridis_r', extend='both')
fig.colorbar(cf, ax=a, label=r'$\log_{10}$ outage', ticks=range(-6, 1))
a.plot(0, 0, '^', ms=11, color='orange', mec='k', label='Source')
a.plot(*users.T, 'o', ms=7, mfc='white', mec='k', label='Users')
a.plot(*ris[:2], 's', ms=9, color='lime', mec='k', label='RIS')
a.plot(xo, yo, '*', ms=15, color='red', mec='white', label='Optimal UAV')
a.plot(*cl, 'X', ms=10, color='red', mec='white', label='Above-cluster')
a.set(xlabel=r'$x_d$ (m)', ylabel=r'$y_d$ (m)', title=r'(a) Outage map at $h_d$ = 222 m, M = 32, 20 dB'); a.grid(False)
a.legend(loc='lower left', fontsize=7)
for M, c in [(0, 'tab:gray'), (32, 'tab:red')]:
    b.semilogy(hs, [pout((xo, yo, h), 20, M) for h in hs], '-', color=c, label=fr'UAV at $(x^\star,y^\star)$, M = {M}')
    b.semilogy(hs, [pout((*cl, h), 20, M) for h in hs], '--', color=c, label=f'Above cluster, M = {M}')
b.set(xlabel=r'UAV altitude $h_d$ (m)', ylabel='Average outage probability', title='(b) Effect of altitude, 20 dB')
b.legend(loc='lower left', fontsize=7.5); fig.tight_layout(); fig.savefig('figures/fig6_placement.png'); plt.show()
print('Best above-cluster outage (M=32, 20 dB): %.3e (paper 4.98e-4)' % min(pout((*cl, h), 20, 32) for h in np.linspace(50, 300, 501)))


# ============================================================================
# ## Step 15 — Table 3: average outage for different UAV placements (fair scheduling)
# ============================================================================

hfine = np.linspace(50, 300, 501)
print(f"{'M':>3}{'rho':>5}  {'Optimised (x,y,h)':22s}{'Optimised':>11}{'Midpoint,100m':>15}{'Above cl.,100m':>16}{'Above cl.,best h':>18}")
for (M, rho), (x, po) in placements.items():
    mid = pout((400, 0, 100), rho, M); c100 = pout((620, 0, 100), rho, M)
    cb = min(pout((620, 0, h), rho, M) for h in hfine)
    print(f"{M:>3}{rho:>5}  ({x[0]:.0f}, {x[1]:.0f}, {x[2]:.0f}){'':8s}{po:>11.3g}{mid:>15.3g}{c100:>16.3g}{cb:>18.3g}")
po32, po0 = placements[(32, 20)][1], placements[(0, 20)][1]
print(f"\nM=32, 20 dB: {pout((400,0,100),20,32)/po32:.0f}x lower than midpoint (paper 166x), "
      f"{min(pout((620,0,h),20,32) for h in hfine)/po32:.0f}x lower than best above-cluster (paper 87x)")
print(f"M=0,  20 dB: gain over best above-cluster = {min(pout((620,0,h),20,0) for h in hfine)/po0:.1f}x (paper ~1.3x)")


# ============================================================================
# ## Step 16 — Fig. 7 (ASER, BPSK) and Fig. 8 (ergodic throughput)
# ============================================================================

fig, ax = plt.subplots(figsize=(5.6, 4.2))
for M in (0, 16, 32):
    ax.semilogy(rho_line, [aser(P_OPT, r, M) for r in rho_line], color=C_M[M], label=f'M = {M} (analysis)')
    v = np.array([mc(P_OPT, r, M, n=150_000, scheds=('fair',))['fair']['ser'] for r in rho_mc])
    ax.semilogy(rho_mc, v, 'o', mfc='none', color=C_M[M])
ax.plot([], [], 'ko', mfc='none', label='Monte Carlo')
ax.set(ylim=(1e-8, 0.5), xlim=(4, 31), xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='ASER (BPSK)')
ax.legend(loc='lower left'); fig.savefig('figures/fig7_aser.png'); plt.show()
for M, ref in [(0, 1.92e-2), (16, 5.91e-4), (32, 1.20e-5)]:
    print(f'M={M:2d}: ASER(20 dB) = {aser(P_OPT, 20, M):.3e} (paper {ref})')

cases = [('Fair, M = 64', 64, 'fair', 'tab:green', '-'), ('Max-SNR, M = 32', 32, 'max', 'tab:blue', '--'),
         ('Fair, M = 32', 32, 'fair', 'tab:red', '-'), ('Round robin, M = 32', 32, 'rr', 'tab:gray', '-.'),
         ('Fair, no RIS', 0, 'fair', 'k', ':')]
cache = {M: {r: mc(P_OPT, r, M, n=150_000) for r in rho_mc} for M in (0, 32, 64)}
fig, ax = plt.subplots(figsize=(5.6, 4.2))
for lab, M, s, c, ls in cases:
    ax.plot(rho_mc, [cache[M][r][s]['C'] for r in rho_mc], ls, marker='o', ms=4, color=c, label=lab)
ax.set(xlabel=r'Transmit SNR $\rho$ (dB)', ylabel='Ergodic throughput (bps/Hz)'); ax.legend(loc='upper left')
fig.savefig('figures/fig8_ergodic.png'); plt.show()
for lab, M, s, *_ in cases:
    print(f'{lab:22s} C(20 dB) = {cache[M][20.0][s]["C"]:.3f} bps/Hz')
print('Paper: Fair M=64 2.650 | Max M=32 2.345 | Fair M=32 2.094 | RR M=32 1.841 | Fair no RIS 1.057')


# ============================================================================
# ## Step 17 — Download all figures (Colab)
# ============================================================================

import shutil
shutil.make_archive('FAIR_RIS_UAV_figures', 'zip', 'figures')
try:
    from google.colab import files; files.download('FAIR_RIS_UAV_figures.zip')
except ImportError:
    print('Saved FAIR_RIS_UAV_figures.zip')
```
